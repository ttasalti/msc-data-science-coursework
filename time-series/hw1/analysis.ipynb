```r
install.packages("readr")
library(readr)

data=read_csv("World_population(1951-2020).csv")
head(data)
str(data)
```

```r
install.packages("dplyr")
library(dplyr)
#data['World Population']=log(data["World Population"])
# Arrange the data by the "Year" column in ascending order
data_sorted <- arrange(data, Year)
library(dplyr)

data_sorted <- arrange(data, Year)
data_sorted <- data_sorted[-c(1:20), ]
# Extract the "World Population" column and create a time series object
datats <- ts(data_sorted$`World Population`,start=c(1971))
data_sorted
```

```r

# Plot the time series
plot.ts(datats)
```

```r
datatsdiff1=diff(datats,differences=1)
plot.ts(datatsdiff1)
```

```r
acf(datatsdiff1,lag.max=20)
```

```r
pacf(datatsdiff1)
```

```r
datatsdiff2=diff(datats,differences=2)
plot.ts(datatsdiff2)
```

```r
acf(datatsdiff2,lag.max=25)
```

```r
pacf(datatsdiff2)
```

```r
arima1=arima(datats,order=c(2,2,0))
arima1
```

```r
library(lmtest)
```

```r
coeftest(arima1)
```

```r
tsdiag(arima1)
```

```r
Box.test(arima1$residuals,lag=6)
```

```r
Box.test(arima1$residuals,lag=12)
```

```r
Box.test(arima1$residuals,lag=18)
```

```r
Box.test(arima1$residuals,lag=24)
```

```r
plot.ts(arima1$residuals)
```

```r
shapiro.test(arima1$residuals)
```

```r
forecast(arima1,h=5)
```

```r
install.packages("bestNormalize")
library(bestNormalize)
```

```r
deneme=bestNormalize(arima1$residuals)
deneme
deneme2=predict(deneme)
hist(deneme2)
plot.ts(deneme2)
```

```r
datats2=predict(deneme,newdata=datats)
plot.ts(datats2)
```
