```r
km_res <- kmeans(scaled_pca, 7, nstart=25) 

fviz_cluster(km_res, data = scaled_pca,
             ellipse.type = "convex", # Concentration ellipse
             star.plot = TRUE, # Add segments from centroids to items
             repel = TRUE, # Avoid label overplotting (slow)
             ggtheme = theme_minimal()
)

```

```r
# Küme merkezlerini alın
centers <- km_res$centers

# Her bir veri noktasının kümeye olan uzaklığının karelerini hesaplayın
distances <- apply(data, 1, function(x) min(apply(centers, 1, function(c) sum((x - c)^2))))

# Her bir kümeye ait WCSS hesaplayın
within_ss <- sum(distances)

# Her bir veri noktasının tüm merkezlere olan uzaklığının karelerini hesaplayın
all_distances <- apply(data, 1, function(x) sum((x - centers[km_res$cluster, ])^2))

# Toplam WCSS hesaplayın
total_ss <- sum(all_distances)

# Küme arası varyansı hesaplayın
between_ss <- total_ss - within_ss

# Kümeleme modelini oluştururken kaç kez "build" adımının çalıştığını bulun
build <- km_res$iter

# Küme merkezlerinin yer değiştirmesi (swap) sayısını bulun
# Bir veri noktası, bir kümeden başka bir kümeye taşındığında, swap işlemi gerçekleşmiş olur
swap <- km_res$tot.withinss - within_ss

# Sonuçları ekrana yazdırın
cat("Within cluster sum of squares (WCSS):", round(within_ss, 5), "\n")
cat("Between_SS:", round(between_ss, 5), "\n")
cat("Total_SS:", round(total_ss, 5), "\n")
cat("(between_SS / total_SS =", round(between_ss / total_ss * 100, 1), "%)\n")
cat("Build:", build, "\n")
cat("Swap:", swap, "\n")
```

```r
# Kümeleme sonuçlarından küme numaralarını alarak bir veri çerçevesi oluşturma
cluster_data <- data.frame(Country = rownames(scaled_pca), Cluster = km_res$cluster)

# Küme numarasına göre veri çerçevesini sıralama
cluster_data <- cluster_data[order(cluster_data$Cluster), ]

# Veri çerçevesini yazdırma
print(cluster_data)
```

```r
# Küme merkezlerine olan uzaklık
distances <- apply(scaled_pca, 1, function(x) min(sapply(1:3, function(i) sum((x - medoids[i,])^2))))

# Her kümeye ait WCSS
wcss_by_cluster <- tapply(distances, pam_data$clustering, sum)

# Toplam WCSS
total_wcss <- sum(distances)

# Toplam veri seti varyansı
total_variance <- sum((scaled_pca - colMeans(scaled_pca))^2)

# Küme arası varyansı
between_ss <- total_variance - total_wcss

# Toplam WCSS ve oranını hesaplama
percentage <- (between_ss / total_variance) * 100

# Sonuçları yazdırma
cat("Within cluster sum of squares by cluster:\n")
cat(paste(round(wcss_by_cluster, 5), "\n"))

cat("Between_SS:", round(between_ss, 5), "\n")
cat("Total_SS:", round(total_variance, 5), "\n")
cat("(between_SS / total_SS =", round(percentage, 1), "%)\n")
```
